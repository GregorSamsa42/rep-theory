```python
#only allow simple modules with drinfeld poly roots in q^i for i nonnegative
k.<q> = PolynomialRing(ZZ)

def laurent_ring(n, base=ZZ):
    """Return Z[y_0^±1, ..., y_{n-1}^±1]"""
    return LaurentPolynomialRing(base, ['y%d' % i for i in range(n)])

# Work in C_l
l = 15
R = laurent_ring(2*l);
y = R.gens()  # y = (y0, y1, ..., y9)

# take as input Drinfeld poly of a simple module, given as a list of its roots, where we write i for q^i
drin_poly = [2,2,4,0,6];
for i in drin_poly:
    if i%2 == 1 or i > 2*l:
        print("The given simple is not in C_l")
        break

def remove_ap2_from_first(lst):
    """
    Find the longest subsequence starting at lst[0]
    where each next element is exactly lst[0] + 2k for increasing k
    """

    if lst==[]:
        return []

    target = lst[0]
    a = target
    seg_length = 1
    indices_to_remove = {0}

    for i in range(1, len(lst)):
        if lst[i] == target + 2:
            indices_to_remove.add(i)
            target += 2
            seg_length += 1
    seg = [seg_length,a]
    # print(segments)
    return ([x for i, x in enumerate(lst) if i not in indices_to_remove],seg)

def char_rk(k,n):
    # computes the q-character of a KR-module W_{k,q^n}
    if k == 0:
        return 1
    if k == 1:
        return y[n] + y[n+2]^(-1)
    return R((char_rk(k-1,n)*char_rk(k-1, n+2) - 1)*(char_rk(k-2,n+2))^(-1))

def qchar(roots):
    segments = []
    while roots:
        # print(segments)
        (roots,new_seg) = remove_ap2_from_first(roots)
        segments.append(new_seg)
    result = 1;
    for [k,n] in segments:
        result *= char_rk(k,n)
    return result
print(char_rk(2,0))

#test = qchar(drin_poly);
#print(test);

def maximal_positive_monomial(f):
    """
    Among all monomials of f with all non-negative exponents,
    return one that is maximal under divisibility:
    no other all-positive monomial n has n/m also all-positive (i.e. m | n).
    """
    d = f.dict()  # exponent_tuple -> coefficient
    
    # collect monomials with all non-negative exponents
    pos_mons = [exp for exp in d.keys() if all(e >= 0 for e in exp)]
    
    if not pos_mons:
        return None  # no all-positive monomials
    
    def divides(a, b):
        """True if monomial a divides monomial b, i.e. b/a is all non-negative."""
        return all(b[i] >= a[i] for i in range(len(a)))
    
    # keep only monomials not properly divided by any other positive monomial
    maximals = []
    for m in pos_mons:
        if not any(divides(m, n) and n != m for n in pos_mons):
            print({m:1})
            return R({m: 1})
    
def monomial_to_indices(m):
    """
    Given a monomial in the LaurentPolynomialRing with non-negative exponents,
    return a list of generator indices with multiplicity.
    E.g. y0^2 * y3 -> [0, 0, 3]
    """
    exp = list(m.dict().keys())[0]  # single exponent tuple
    return [i for i, e in enumerate(exp) for _ in range(e)]
    
# print(monomial_to_indices(maximal_positive_monomial(test)[0]))
# give a tensor product V_(n_i)(q^a_i) as a list
test_tp = [[1,0],[1,0],[1,0]];
def comp_factors(tp):
    # compute q-character
    character = 1
    for ev_mod in test_tp:
        character *= char_rk(ev_mod[0],ev_mod[1])
    print(character)
    while character != 0:
        simple = maximal_positive_monomial(character);
        print(monomial_to_indices(simple));
        print(qchar(monomial_to_indices(simple)))
        character -= qchar(monomial_to_indices(simple));
        #print(character)
#print(qchar([0,2]))
comp_factors(test_tp)
```
